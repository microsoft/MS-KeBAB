```python
import json
import pathlib

import pandas as pd
import plotly.express as px
```

Sample element from the dataset
---

```python
# load the Wikidata person names dataset
dataset_path = pathlib.Path.home() / "OneDrive - Microsoft" / "Benchmark" / "Datasets" / "Wikidata" / "Person Names" / "2024-06-27" / "wikidata_person_names.jsonl"

with open(dataset_path, encoding="utf-8") as f:
    entities = [json.loads(line.strip()) for line in f]

# convert the entities to a DataFrame
df = pd.DataFrame(entities)

print("Sample entity:")
entities[42]
```

Dataset statistics
---

```python
print(f"Number of entities: {len(df):,d}")

# average and maximum number of aliases per entity
df["aliases_count"] = df["aliases"].apply(len)
print(f"Average number of aliases per entity: {df['aliases_count'].mean():.2f}")
print(f"Maximum number of aliases per entity: {df['aliases_count'].max()}")
print(f"Number of entities with more than 10 aliases: {df['aliases_count'].gt(10).sum()}")
print(f"Maximum number of aliases per entity: {df['aliases_count'].max()}")
```

```python
# histogram
fig = px.histogram(x=df["aliases_count"].clip(upper=10), nbins=20, title="Entities by # Aliases",
                   labels={"x": "number of aliases", "y": "# entities"})

total_entities = len(df)
for i in sorted(df["aliases_count"].unique())[:9]:
    count = (df["aliases_count"] == i).sum()
    fig.add_annotation(x=i, y=count, text=f"{count/total_entities:.1%}",
                       showarrow=False, yshift=10)

fig.update_layout(xaxis=dict(nticks=10))
fig.update_yaxes(title_text="entity count")
fig.update_layout(width=1000)
fig.show()
```

Entities with the most aliases
---

```python
# top entities with the most aliases
df.sort_values("aliases_count", ascending=False)[["id", "name", "aliases_count", "aliases"]].head(10)
```

Entities with the longest aliases
---

```python
# top entities with the longest aliases
df["longest_alias"] = df["aliases"].apply(lambda aliases: max(aliases, key=len))
df["longest_alias_length"] = df["longest_alias"].apply(len)
df.sort_values("longest_alias_length", ascending=False)[['id', 'name', 'longest_alias', 'longest_alias_length']].head(10)
```
